# Annual Employment by Industry

## Dataset Description
This dataset contains the **annual number of employed persons broken down by industry sector**, from the DOSM Labour Force Survey (LFS). While this does not contain wage figures directly, it shows **which sectors absorb the most workers** — a proxy for graduate employment demand by field of study.

> **Note:** The original plan was to use the DOSM Salaries & Wages Survey (industry x sex breakdown). That dataset was hosted on an old Google Cloud Storage bucket (`storage.googleapis.com/dosm-public-economy`) which is no longer publicly accessible. This dataset is the closest available replacement on the current `storage.dosm.gov.my` infrastructure.
>
> For salary data by industry, the best current alternative is the **Salaries & Wages Survey Report PDF** available at: https://storage.dosm.gov.my/labour/salaries_wages_2024.pdf

## Data Source
| Field | Detail |
|---|---|
| **Publisher** | Department of Statistics Malaysia (DOSM) |
| **Survey** | Labour Force Survey (LFS), annual |
| **Catalogue** | https://open.dosm.gov.my/data-catalogue/lfs_year_emp_industry |
| **Parquet URL** | `https://storage.dosm.gov.my/labour/lfs_year_emp_industry.parquet` |
| **CSV URL** | `https://storage.dosm.gov.my/labour/lfs_year_emp_industry.csv` |
| **License** | CC BY 4.0 — DOSM |
| **Coverage** | 2010 – 2023 |

## Column Descriptions
| Column | Type | Description |
|---|---|---|
| `date` | date | Year of survey |
| `industry` | string | Industry classification (MSIC 2008) |
| `employed` | float | Number of employed persons ('000) |

## Relevance to EduNilai
Shows labour market absorption by sector — Engineering, Manufacturing, Finance, Education, and Health sectors can be mapped to degree fields to estimate employment odds for each graduate type.

```python
# pip install pandas fastparquet
import pandas as pd
import matplotlib.pyplot as plt
import os

URL = 'https://storage.dosm.gov.my/labour/lfs_year_emp_industry.parquet'
df = pd.read_parquet(URL)
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])

df.head()
```

```python
print("Shape:", df.shape)
print("\nColumn dtypes:")
print(df.dtypes)
print("\nUnique industries:", sorted(df['industry'].unique()))
print("\nYear range:", df['date'].dt.year.min(), "–", df['date'].dt.year.max())
```

```python
# Filter 2010 onwards
df_2010 = df[df['date'].dt.year >= 2010].copy()

# Latest year snapshot
latest_year = df_2010['date'].max()
df_latest = df_2010[df_2010['date'] == latest_year].sort_values('employed', ascending=False)

print(f"Latest year: {latest_year.year}")
print(df_latest[['industry', 'employed']].to_string(index=False))
```

```python
# Bar chart: employment by industry (latest year)
fig, ax = plt.subplots(figsize=(10, 6))
df_plot = df_latest.sort_values('employed', ascending=True)
ax.barh(df_plot['industry'], df_plot['employed'], color='steelblue')
ax.set_xlabel("Employed Persons ('000)")
ax.set_title(f"Employment by Industry ({latest_year.year})")
plt.tight_layout()
plt.show()
```

```python
os.makedirs('../data/salary', exist_ok=True)
df_2010.to_csv('../data/salary/employment_by_industry.csv', index=False, encoding='utf-8')
print("Saved -> ../data/salary/employment_by_industry.csv")
print(f"Rows: {len(df_2010)}")
```
